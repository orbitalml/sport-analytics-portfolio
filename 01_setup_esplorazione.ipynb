```python
def risultato_partita (gol_casa, gol_trasferta):
    """ Return match result (1,x,2) based on the goal
    of each team
    >>> risultato_partita (1,0)
    1
    >>> risultato_partita (3,3)
    X
    >>> risultato_partita (1,2)
    2
    """
    if gol_casa > gol_trasferta:
        return "1"
    elif gol_casa == gol_trasferta:
        return "X"
    else:
        return "2"

print(risultato_partita(2, 1))
print(risultato_partita(0, 0))
print(risultato_partita(1, 3))
```

```python
partite = [
    ("Inter", "Milan", 2, 1),
    ("Napoli", "Juventus", 1, 1),
    ("Roma", "Lazio", 0, 2),
    ("Atalanta", "Fiorentina", 3, 0),
    ("Como", "Verona", 1, 2)
]

for match in partite:
    print (match[0], " vs ",match[1], " -> ", risultato_partita(match[2],match[3])) 
```

```python
casa = 0
trasferta = 0
pareggi = 0

for match in partite:
    if risultato_partita(match[2],match[3]) == "1":
        casa=casa+1
    elif risultato_partita(match[2],match[3]) == "2":
        trasferta=trasferta+1
    else:
        pareggi=pareggi+1
        
print ("Percentuale vittorie in casa: ", casa/(casa+trasferta+pareggi)*100,"%")
print ("Percentuale vittorie in trasferta: ", trasferta/(casa+trasferta+pareggi)*100,"%")
print ("Percentuale pareggi: ", pareggi/(casa+trasferta+pareggi)*100,"%")
```
